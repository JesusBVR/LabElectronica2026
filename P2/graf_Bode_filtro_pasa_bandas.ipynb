```python
import numpy as np
import matplotlib.pyplot as plt

## Función pasa bandas
def model(s, w1, w2):
	fucTransLP = w2 / (s+w2)
	funcTransHP = s / (s+w1)

	return funcTransHP * fucTransLP

## Difine el rango de frecuencias y el complejo s
sReal = np.logspace(0, 5, num=1000)
s = 1j * sReal

## Define las frecuencias de corte 1 para HP, 2 para  LP
w1 = 60
w2 = 3100

## Tranformación ganancia lineal a decibeles
model_db = 20*np.log10(np.abs(model(s, w1, w2)))

## Grafica de ganancia y frec de corte a -3dB
plt.plot(np.abs(s), model_db)
plt.axhline(y=-3, ls='--', c='r')
plt.axvline(x=60, c='g')
plt.axvline(x=3100, c='g')
plt.grid(visible=True)
plt.xscale('log')
plt.xlim((10e-1, 10e4))
plt.ylim((-50, 0))
plt.show()
```
