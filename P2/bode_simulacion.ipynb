```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

path = 'C:\\users\\Jesus-VR\\Desktop\\data.csv'
path2 = 'C:\\users\\Jesus-VR\\Desktop\\data2.csv'

data = pd.read_csv(path, delimiter=';',usecols=(0,1)).to_numpy()
data2 = pd.read_csv(path2, delimiter=';',usecols=(0,1)).to_numpy()

frec = data[:,0]
gan = data[:,1]

frec2 = data2[:,0]
gan2 = data2[:,1]

# print(np.max(gan))
plt.semilogx(frec, gan)
current_ticks = np.arange(-10, 2, 2)
all_ticks = np.sort(np.append(current_ticks, np.max(gan)))
plt.yticks(all_ticks)
plt.axhline(y=np.max(gan))
plt.show()

plt.semilogx(frec2, gan2)
current_ticks = np.arange(-10, 2, 2)
all_ticks = np.sort(np.append(current_ticks, np.max(gan2)))
plt.yticks(all_ticks)
plt.axhline(y=np.max(gan2))
plt.show()
```
