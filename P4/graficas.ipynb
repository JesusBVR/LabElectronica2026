## Transformador

```python
import pandas as pd
import matplotlib.pyplot as plt

transfo1 = pd.read_csv('.\\datos\\IN-TRANSFO-NEGATIVO.csv', sep=';', usecols=[1])
transfo2 = pd.read_csv('.\\datos\\IN-TRANSFO-POSITIVO.csv', sep=';', usecols=[1])
time = pd.read_csv('.\\datos\\IN-TRANSFO-POSITIVO.csv', sep=';', usecols=[0])

plt.plot(time, transfo2)
plt.yticks([x for x in range(-30, 40, 10)]+[33.9399, -33.9399])
plt.title('Tensión en la primer linea del devanado secundario', weight='bold', pad=15)
plt.xlabel('t [s]')
plt.ylabel('V(t) [V]')
plt.grid(axis='y')
plt.tight_layout()
plt.show()

plt.plot(time, transfo1)
plt.yticks([x for x in range(-30, 40, 10)]+[33.9399, -33.9399])
plt.title('Tensión en la segunda linea del devanado secundario', weight='bold', pad=15)
plt.xlabel('t [s]')
plt.ylabel('V(t) [V]')
plt.grid(axis='y')
plt.tight_layout()
plt.show()
```

## Capacitor

```python
import pandas as pd
import matplotlib.pyplot as plt

C1 = pd.read_csv('.\\datos\\OUT-C-NEGATIVO.csv', sep=';', usecols=[1])
C2 = pd.read_csv('.\\datos\\OUT-C-POSITIVO.csv', sep=';', usecols=[1])
time = pd.read_csv('.\\datos\\OUT-C-NEGATIVO.csv', sep=';', usecols=[0])

plt.plot(time, C1)
plt.yticks([x for x in range(-30, 10, 10)]+[-33.1814])
plt.title('Tensión en el capacitor de salida negativa', weight='bold', pad=15)
plt.xlabel('t [s]')
plt.ylabel('V(t) [V]')
plt.grid(axis='y')
plt.tight_layout()
plt.show()

plt.plot(time, C2)
plt.yticks([x for x in range(0, 40, 10)]+[33.1814])
plt.title('Tensión en el capacitor de salida positiva', weight='bold', pad=15)
plt.xlabel('t [s]')
plt.ylabel('V(t) [V]')
plt.grid(axis='y')
plt.tight_layout()
plt.show()
```
