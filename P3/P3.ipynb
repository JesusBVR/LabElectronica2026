# ANÁLISIS 1N4007

## Graficas

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt


path = '.\\datos\\'

data = pd.read_csv(path+'datos_1N4007_experimental.csv')
V_D_sim = pd.read_csv(path+'VD_simulacion.csv', sep=';', usecols=[0,1])
I_D_sim = pd.read_csv(path+'ID_simulacion.csv', sep=';', usecols=[1])*1000


plt.plot(data['Vs'], data['Vd'], label='Datos Experimentales')
plt.plot(V_D_sim['Vs'], V_D_sim['VD'], label='Datos Simulados', ls='--')
plt.xlabel(r'$V_S$'+' [V]')
plt.ylabel(r'$V_D$'+' [V]')
plt.title(r'Comparación experimetal y teórica $V_D$ vs $V_S$'+'\n'+'1N4007')
plt.legend()
plt.show()

plt.plot(data['Vd'], data['ID'], label='Datos Experimentales')
plt.plot(V_D_sim['VD'], I_D_sim, label='Datos Simulados', ls='--')
plt.xlabel(r'$V_D$'+' [V]')
plt.ylabel(r'$I_D$'+' [mA]')
plt.title(r'Comparación experimental y teórica $I_D$ vs $V_D$'+'\n'+'1N4007')
plt.legend()
plt.show()
```

## Tabla de datos en latex

```python
tabla = data.drop(columns=['VR Prac']).style.hide(axis='index')
tabla = tabla.format({
	'Vs':'{:.1f}',
	'Vd':'{:.4f}',
	'VR Teo':'{:.3f}',
	'ID':'{:.2f}'
	})
tabla = tabla.to_latex(
	caption='Caracterización del diodo 1N4007.',
	label='tab:dato_1N4007',
	position='htbp',
	hrules=True
)
print(tabla)
```

# ANÁLISIS 1N4733A

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

path = '.\\datos\\'

data_exp = pd.read_csv(path+'datos_1N4733A_experimental.csv')
data_sim = pd.read_csv(path+'1n4733a_simulacion.csv', sep=';')

plt.plot(data_exp['Vs'], data_exp['Vout'], label='Datos Experimentales')
plt.plot(data_sim['Vs'], data_sim['Vout'], ls='--', label='Datos simulados')
plt.xlabel(r'$V_s$ [V]')
plt.ylabel(r'$V_{out}$ [V]')
plt.axhline(y=5.1, c="#00A708", ls=':', label=r'$V_z=5.1$ V')
plt.title(r'Comparación experimental y teórica $V_{out}$ vs $V_s$'+'\n'+'1N4733A')
plt.legend()
# plt.xticks(data_exp['Vs'])
plt.show()
```

## Tabla de datos

```python
Vs2loc = (0.0, 3.0,5.0,5.5,6.5,8.0)
Vout_dic = {'Vs':[],'Vout Teo':[]}

for Vs in Vs2loc:
	Vout = data_sim.loc[data_sim['Vs']==Vs, 'Vout'].iloc[0]
	Vout_dic['Vs'].append(Vs)
	Vout_dic['Vout Teo'].append(Vout)
Vout_dic['Vout Exp'] = data_exp['Vout']
Vout_table = pd.DataFrame(Vout_dic).style.hide(axis='index')
Vout_table = Vout_table.format({
	'Vs':'{:.1f}',
	'Vout Teo':'{:.3f}',
	'Vout Exp':'{:.3f}'
})
Vout_table = Vout_table.to_latex(
	caption='Caracterización del diodo 1N4733A.',
	label='tab:dato_1N4733A',
	position='htbp',
	hrules=True
)

print(Vout_table)
```
