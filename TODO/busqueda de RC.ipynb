```python
import numpy as np
import pandas as pd
```

```python
R = np.array([1,2,3,4])
C = np .array([1,2])

multiplicador = np.array([10**i for i in range(1,7)])

R_t = np.outer(R,multiplicador).reshape(-1)
tau = np.outer(R_t,C).reshape(-1)
```

```python
tauTeo = 5
# tauTeo = int(input("Escribe el valor de Tau teórico: \n"))

diferencia = (tau - tauTeo)
orden = np.argsort(diferencia)
diferencia[orden]
df = pd.DataFrame({'Tau Calculado':tau[orden], 'Diferencia':diferencia[orden]})
df
```
